```python
# 1.Load Data & Overview
print("\n1️⃣ OVERVIEW & SCHEMA")
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
print("\nColumn Data Types:")
print(df.dtypes)
```

```python
print("\n" + "=" * 50)
print("⚠️ 2. MISSING VALUES AUDIT")
print("=" * 50)
```

```python
# 2. Missing Values
missing_counts = df.isna().sum()
total_missing = missing_counts.sum()

if total_missing == 0:
    print("✅ No missing values found!")
else:
    for col in df.columns:
        count = missing_counts[col]
        if count > 0:
            percentage = (count / len(df)) * 100
            print(f"• Column '{col}': {count} missing values ({percentage:.1f}%)")
```

```python
print("\n" + "=" * 50)
print("🔁 3. DUPLICATE RECORDS AUDIT")
print("=" * 50)
```

```python
# 3. Check Duplicates & Anomalies
duplicate_rows = df.duplicated().sum()
duplicate_ids = df["Student_ID"].duplicated().sum()

print(f"Full Duplicate Rows: {duplicate_rows}")
print(f"Duplicate Student_IDs: {duplicate_ids}")

if duplicate_ids > 0:
    print("\nRepeated Student IDs found:")
    print(df[df["Student_ID"].duplicated(keep=False)][["Student_ID", "Full_Name"]])
```

```python
print("\n" + "=" * 50)
print("🔍 4. DATA ANOMALIES DETECTED")
print("=" * 50)
```

```python
# Check Age anomalies (Negative ages or extreme values like > 100)
bad_age = df[(df["Age"] < 0) | (df["Age"] > 100)]
if len(bad_age) > 0:
    print(f"⚠️ Found {len(bad_age)} row(s) with invalid Age values:")
    print(bad_age[["Student_ID", "Full_Name", "Age"]])
```
